### 2. 평균 나이/월드컵 경험과 공격/수비 지수 상관관계
- 평균나이 또는 선수단의 평균 월드컵 본선 진출 횟수가 공격/수비에 강점을 가지는지 시각화

#### 평균 나이와 공격/수비 지수 상관관계

```python
fig, ax = plt.subplots(figsize=(14, 6), ncols=2, nrows=1)

sns.regplot(data=df, x='Avg_Age', y='ATK_INDEX', ax=ax[0])
ax[0].set_title('평균 나이 vs 공격 지수')
ax[0].set_xlabel('평균 나이')
ax[0].set_ylabel('공격 지수')
ax[0].grid(True)

sns.regplot(data=df, x='Avg_Age', y='DEF_INDEX', ax=ax[1], color='darkorange')
ax[1].set_title('평균 나이 vs 수비 지수')
ax[1].set_xlabel('평균 나이')
ax[1].set_ylabel('수비 지수')
ax[1].grid(True)

plt.show()
```

- 회귀선이 거의 수평에 가깝고 데이터가 넓게 분포되어 있으므로 평균 나이가 공격 지수나 수비 지수를 예측하는 데 유의미한 영향을 미치지 않는다는 것을 확인할 수 있다.

#### 월드컵 경험과 공격/수비 지수 상관관계

```python
fig, ax = plt.subplots(figsize=(14, 6), ncols=2, nrows=1)

sns.regplot(data=df, x='Avg_Apps', y='ATK_INDEX', ax=ax[0])
ax[0].set_title('월드컵 경험 vs 공격 지수')
ax[0].set_xlabel('평균 월드컵 본선 진출 횟수')
ax[0].set_ylabel('공격 지수')
ax[0].grid(True)


sns.regplot(data=df, x='Avg_Apps', y='DEF_INDEX', ax=ax[1], color='darkorange')
ax[1].set_title('월드컵 경험 vs 수비 지수')
ax[1].set_xlabel('평균 월드컵 본선 진출 횟수')
ax[1].set_ylabel('수비 지수')
ax[1].grid(True)

plt.show()
```

- 공격 지수와의 관계에서 회귀선이 상승 그래프를 그리고 있으므로 선수들의 평균 월드컵 본선 진출 횟수가 많을수록 팀의 공격력이 강해진다는 것을 알 수 있다. 수비 지수와의 관계에서도 상승 그래프를 그리고 있으나 공격 지수만큼 관계가 강하게 나타나지는 않는다. 

```python
atk_def_df = df[['Avg_Age', 'Avg_Apps', 'ATK_INDEX', 'DEF_INDEX']]

plt.figure(figsize=(8, 6))
corr_matrix = atk_def_df.corr(numeric_only=True)

sns.heatmap(corr_matrix, annot=True)
plt.show()
```

- 평균 나이는 공격/수비지수에 거의 영향을 미치지 않고 월드컵 본선 진출 경험은 공격력에 약간이나마 영향을 주는 것을 재확인할 수 있다. 
- 또한 추가로 공격력이 강한 팀이 수비력도 강하다는 것을 확인할 수 있다.

결론: 월드컵 본선이라는 큰 무대에서의 경험은 수비 관련 지표보다 공격 관련 지표들에 더 큰 영향을 미친다는 것을 의미하며 득점 기회를 더 많이 만드는 등 최종 성적과 직결되는 공격 지표에 강점을 보인다는 것을 의미한다. 또한, 팀의 나이보다는 월드컵 경험이 팀의 전력에 더 중요한 영향을 미치는 요소일 가능성이 높다는 것을 의미한다.

### 3. 월드컵 순위와 공격/수비 지수 상관관계 
- 월드컵 순위 상위&하위 나라가 공격/수비에 강점을 가지는지 시각화

#### 월드컵 최종 순위에 따른 공격/수비 지수 상관관계

```python
fig, ax = plt.subplots(figsize=(14, 6), ncols=2, nrows=1)

sns.regplot(data=df, x='Wc_Rank', y='ATK_INDEX', ax=ax[0])
ax[0].set_title('월드컵 순위 vs 공격 지수')
ax[0].set_xlabel('월드컵 순위')
ax[0].set_ylabel('공격 지수')
ax[0].grid(True)

sns.regplot(data=df, x='Wc_Rank', y='DEF_INDEX', ax=ax[1], color='darkorange')
ax[1].set_title('월드컵 순위 vs 수비 지수')
ax[1].set_xlabel('월드컵 순위')
ax[1].set_ylabel('수비 지수')
ax[1].grid(True)

plt.show()
```

```python
plt.figure(figsize=(10,6))
scatter = sns.scatterplot(
    data=df,
    x="ATK_INDEX", 
    y="DEF_INDEX",
    hue="Wc_Rank",
    size="Wc_Point",
    palette="coolwarm",
    sizes=(50, 400),
    alpha=0.7
)

plt.title("공격 vs 수비 (최종 순위)")
plt.xlabel("공격 지수")
plt.ylabel("수비 지수")
plt.grid(True)
plt.legend()
plt.show()
```

- 성적에 따라 수비/공격 지수가 유의미하게 분리되지 않고 골고루 섞여 보인다. 공격 지수와 수비 지수가 모두 높은 팀일수록 월드컵에서 좋은 성적을 거둘 가능성이 높아지는 경향은 있지만 예외가 있음을 확인할 수 있다.

#### 진출 라운드에 따른 공격/수비 지수 상관관계

```python
def rank_group(rank):
    if rank <= 4:
        return "준결승 이상 진출"
    elif rank <= 16:
        return "16강 이상 진출"
    else:
        return "조별 리그 탈락"

df['Rank_Group'] = df['Wc_Rank'].apply(rank_group)

fig = px.scatter(
    df,
    x="ATK_INDEX",
    y="DEF_INDEX",
    color="Rank_Group",
    size="Wc_Point",    
    hover_data=["Year","Nation", "Wc_Rank", "Wc_Point"],
)

fig.update_layout(xaxis_title="공격 지수",yaxis_title="수비 지수",title="공격 vs 수비 지수 (진출 라운드)")
fig.show()
```

#### 연도별 월드컵 진출 라운드에 따른 공격/수비 지수 상관관계

```python
fig = px.scatter(
    df,
    x="ATK_INDEX",
    y="DEF_INDEX",
    color="Rank_Group",
    size="Wc_Point",
    hover_data=["Year","Nation", "Wc_Rank", "Wc_Point"],
    animation_frame="Year",
)

fig.update_layout(xaxis_title="공격 지수", yaxis_title="수비 지수", title="연도별 월드컵 공격 vs 수비 지수 (진출 라운드)")
fig.show()
```

결론: 월드컵의 변수성으로 인해 성적에 따라 수비/공격 지수가 유의미하게 분리되지 않는 것으로 보인다.

### 4. 우승 여부와 평균 나이/월드컵 경험의 상관관계

```python
fig, ax = plt.subplots(figsize=(20,10), ncols=2, nrows=1)

sns.boxplot(ax=ax[0], data=df, x="Wc_Rank", y="Avg_Age")
ax[0].set_title("최종 순위에 따른 평균 나이 분포")
ax[0].set_xlabel("최종 순위")
ax[0].set_ylabel("평균 나이")

sns.boxplot(ax=ax[1], data=df, x="Wc_Rank", y="Avg_Apps")
ax[1].set_title("최종 순위에 따른  평균 월드컵 본선 출전 횟수 분포")
ax[1].set_xlabel("최종 순위")
ax[1].set_ylabel("평균 본선 출전 수")

plt.show()
```

- 최종 순위가 높을수록 선수들의 평균 월드컵 본선 출전 경험이 많고 안정적임을 보여주지만, 평균 나이에 대해서는 순위별로 차이가 뚜렷이 보이지 않는다.
- 아래 시각화에서 분석 단위를 축소하여 준결승 진출 여부로 비교해 보려 한다.

#### 준결승 진출 여부와 평균 나이/평균 출전 횟수 상관관계

```python
df['Wc_Top4'] = df['Wc_Rank'].apply(lambda x: '준결승 진출팀' if x <= 4 else '나머지 팀')

fig, ax = plt.subplots(figsize=(20,10), ncols=2, nrows=1)

sns.boxplot(ax=ax[0], data=df, x="Wc_Top4", y="Avg_Age", palette="RdBu")
ax[0].set_title("평균 나이 비교 (준결승 진출팀 vs 나머지 팀)")
ax[0].set_xlabel("준결승 진출 여부")
ax[0].set_ylabel("평균 나이")

sns.boxplot(ax=ax[1], data=df, x="Wc_Top4", y="Avg_Apps", palette="Set3")
ax[1].set_title("평균 월드컵 본선 출전 횟수 비교 (준결승 진출팀 vs 나머지 팀)")
ax[1].set_xlabel("준결승 진출 여부")
ax[1].set_ylabel("평균 본선 출전 수")

plt.show()
```

- 본선 출전 횟수에 대해서는 준결승 진출팀이 나머지 팀에 비해 선수단 평균 본선 출전 경험이 더 많다는 점에서 차이가 있지만, 평균 나이는 차이가 거의 없다.
- 아래 시각화에서 분석의 단위를 조금 더 축소하여 우승 여부로 비교해 보려 한다.

#### 우승 여부와 평균 나이/평균 출전 횟수 상관관계

```python
df['Wc_Winner'] = df['Wc_Rank'].apply(lambda x: '우승팀' if x == 1 else '나머지 팀')

fig, ax = plt.subplots(figsize=(20,10), ncols=2, nrows=1)

sns.boxplot(ax=ax[0], data=df, x="Wc_Winner", y="Avg_Age", palette="RdBu")
ax[0].set_title("평균 나이 비교 (우승팀 vs 나머지 팀)")
ax[0].set_xlabel("우승 여부")
ax[0].set_ylabel("평균 나이")

sns.boxplot(ax=ax[1], data=df, x="Wc_Winner", y="Avg_Apps", palette="Set3")
ax[1].set_title("평균 월드컵 본선 출전 횟수 비교 (우승팀 vs 나머지 팀)")
ax[1].set_xlabel("우승 여부")
ax[1].set_ylabel("평균 본선 출전 수")

plt.show()
```

- 우승팀의 평균 나이가 나머지 팀들보다 전반적으로 낮게 분포하고 있다는 것을 알 수 있다. 이는 우승팀이 상대적으로 젊은 선수들로 구성되어 많은 활동량과 강한 체력을 바탕으로 경기를 이끌어가며 우승에 도달하게 된다고 해석할 수 있다.

- 우승팀 선수들의 평균 본선 진출 경험이 나머지 팀 선수들보다 더 많음을 확인할 수 있다. 이는 우승을 위해서는 큰 무대에서 압박감을 이겨내고 경기를 운영할 수 있는 노련함이 중요하다고 해석할 수 있다.

결론: 우승팀은 상대적으로 젊은 스쿼드를 가졌지만 월드컵 본선 진출 경험은 더 풍부한 선수들로 이루어져 있다. 즉, 젊은 에너지와 노련한 경험이 균형을 이룬 팀이라고 할 수 있다. 따라서 우승팀은 월드컵을 치러본 경험이 있는 선수들을 적절히 포함하여 적절한 세대교체를 통해 신구 조화를 이룬 팀일 가능성이 높다.
