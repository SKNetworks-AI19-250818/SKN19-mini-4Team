```python
# Round 순위 매핑
round_rank_map = {
    'Final': ['1위', '2위'],                 # 홈팀 승리 기준
    'Third-place match': ['3위', '4위'],     # 홈팀 승리 기준
    'Semi-finals': ['4강', '4강'],           # 준결승 진출 팀
    'Quarter-finals': ['8강', '8강'],
    'Round of 16': ['16강', '16강'],
    'Second group stage': ['8강', '8강'],    # 구체적 구조에 따라 조정 가능
    'First group stage': ['32강', '32강'],
    'Group stage': ['32강', '32강']
}

# 결과 저장
result = []
for idx, row in df.iterrows():
    ranks = round_rank_map.get(row['Round'], ['Unknown', 'Unknown'])
    # 홈팀
    result.append({'Year': row['Year'], 'Team': row['home_team'], 'Final_Rank': ranks[0]})
    # 원정팀
    result.append({'Year': row['Year'], 'Team': row['away_team'], 'Final_Rank': ranks[1]})

final_df = pd.DataFrame(result)
final_df = final_df.drop_duplicates(subset=['Year', 'Team']).reset_index(drop=True)

print(final_df)
```
