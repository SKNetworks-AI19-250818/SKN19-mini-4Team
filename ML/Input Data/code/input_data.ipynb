```python
fifa_rank_df = pd.read_csv("./../data/source data/avg_rank_points.csv")
fifa_rank_df.head()
```

```python
fifa_rank_df = fifa_rank_df[fifa_rank_df['wc_year'] == 2026]
fifa_rank_df
```

```python
fifa_rank_df[fifa_rank_df['team'] == 'South Korea']
```

```python
columns = {
    'team':'Eng_Nation',
    'rank':'F_Rank',
    'wc_year':'Year',
    'total_points':'F_Point',
    'rank_diff':'F_Rd',
    'points_diff':'F_Pd'
}

renamed_df = fifa_rank_df.rename(columns=columns)
fifa_rank_df = pd.DataFrame(renamed_df)
fifa_rank_df
```

```python
nation_map = {
    'Korea Republic':'South Korea',
    'Korea DPR':'North Korea',
    'USA':'United States',
    'IR Iran':'Iran',
    "Côte d'Ivoire":'Ivory Coast'
}

fifa_rank_df['Eng_Nation'] = fifa_rank_df['Eng_Nation'].str.strip().replace(nation_map)
print(fifa_rank_df['Eng_Nation'].unique())
print(fifa_rank_df['Eng_Nation'].isnull().sum())
```

```python
fifa_rank_df[fifa_rank_df['Eng_Nation'] == 'South Korea']
```

```python
merged_df = pd.merge(input_df, fifa_rank_df, on=['Eng_Nation'], how='left')
merged_df.info()
```

```python
merged_df[merged_df.isna().any(axis=1)]
```

```python
merged_df.to_csv("./../data/result data/input_data.csv", index=False)
merged_df.head()
```

### 2026 FIFA 랭킹 데이터

```python
fifa_rank_24_df = pd.read_csv("./../data/source data/fifa_ranking_2024-09-19.csv")
fifa_rank_24_df.head()
```

```python
fifa_rank_25_df = pd.read_csv("./../data/source data/fifa_ranking_2025-09-18.csv")
fifa_rank_25_df.head()
```

```python
fifa_rank_24_df = fifa_rank_24_df[["Team", "Rank", "Total_Points"]]
fifa_rank_25_df = fifa_rank_25_df[["Team", "Rank", "Total_Points"]]
fifa_rank_24_df, fifa_rank_25_df
```

```python
merged_df = fifa_rank_25_df.merge(fifa_rank_24_df[['Team', 'Rank', 'Total_Points']], on='Team', how='left', suffixes=('_25', '_24'))
merged_df['F_Rd'] = merged_df['Rank_25'] - merged_df['Rank_24']
merged_df['F_Pd'] = merged_df['Total_Points_25'] - merged_df['Total_Points_24']
merged_df
```

```python
merged_df = merged_df[['Team', 'Rank_25', 'Total_Points_25', 'F_Rd', 'F_Pd']]
merged_df
```

```python
columns = {
    'Team':'Eng_Nation',
    'Rank_25':'F_Rank',
    'Total_Points_25':'F_Point',
}

renamed_df = merged_df.rename(columns=columns)
merged_df = pd.DataFrame(renamed_df)
merged_df
```

```python
merged_df[merged_df['Eng_Nation'] == 'Czechia']
```

```python
nation_map = {
    'Korea Republic':'South Korea',
    'Korea DPR':'North Korea',
    'USA':'United States',
    'IR Iran':'Iran',
    "Côte d'Ivoire":'Ivory Coast'
}

merged_df['Eng_Nation'] = merged_df['Eng_Nation'].str.strip().replace(nation_map)
print(merged_df['Eng_Nation'].unique())
print(merged_df['Eng_Nation'].isnull().sum())
```

```python
merged_df = pd.merge(input_df, merged_df, on=['Eng_Nation'], how='left')
merged_df.info()
```

```python
merged_df[merged_df.isna().any(axis=1)]
```

```python
merged_df.to_csv("./../data/result data/annual_input_data.csv", index=False)
merged_df.head()
```
