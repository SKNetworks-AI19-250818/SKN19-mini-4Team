```python
fifa_rank_df = pd.read_csv("./../data/source data/avg_rank_points.csv")
fifa_rank_df.head()
```

```python
fifa_rank_df = fifa_rank_df[fifa_rank_df['wc_year'] == 2026]
fifa_rank_df
```

```python
columns = {
    'team':'Eng_Nation',
    'rank':'F_Rank',
    'wc_year':'Year',
    'total_points':'F_Point',
    'rank_diff':'F_Rd',
    'points_diff':'F_Pd'
}

renamed_df = fifa_rank_df.rename(columns=columns)
fifa_rank_df = pd.DataFrame(renamed_df)
fifa_rank_df
```

```python
nation_map = {
    'Korea DPR':'South Korea',
    'Korea Republic':'North Korea',
    'USA':'United States',
    'IR Iran':'Iran',
    "Côte d'Ivoire":'Ivory Coast'
}

fifa_rank_df['Eng_Nation'] = fifa_rank_df['Eng_Nation'].str.strip().replace(nation_map)
print(fifa_rank_df['Eng_Nation'].unique())
print(fifa_rank_df['Eng_Nation'].isnull().sum())
```

```python
merged_df = pd.merge(input_df, fifa_rank_df, on=['Eng_Nation'], how='left')
merged_df.info()
```
