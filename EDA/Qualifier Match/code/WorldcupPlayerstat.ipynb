```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/source data/world_cup_comparisons.csv')
df
```

```python


# 공격 지수와 수비 지수 컬럼 정의
attack_cols = ['goals_z','xg_z','crosses_z','nsxg_z','fouled_z','boxtouches_z',
               'passes_z','progpasses_z','takeons_z','progruns_z']

defense_cols = ['fouls_z','interceptions_z','clearances_z','tackles_z','blocks_z','aerials_z']

# 모든 시즌, 모든 팀에 대해 평균 계산
team_avg_stats = df.groupby(['season','team']).apply(
    lambda x: pd.Series({
        'attack_index': x[attack_cols].mean().mean(),
        'defense_index': x[defense_cols].mean().mean()
    })
).reset_index()

print(team_avg_stats)
```

```python
team_avg_stats.to_csv('../data/result data/wc_team_stat.csv',index=False,encoding='utf-8')
```
